## Part 1. 리스트와 반복문을 이용한 데이터 처리

여러 개의 데이터를 리스트에 저장하고 `for` 반복문을 이용하여 하나씩 처리합니다.

다음과 같은 작업을 수행합니다.

- 리스트 생성
- 리스트의 데이터 순회
- 조건에 맞는 데이터 찾기
- 조건을 만족하는 데이터 개수 계산
- 계산 결과 출력

**필수 역량**

`list`, `for`, `if`

````python
scores = [90, 85, 45, 70, 97]
total = 0
up_count = 0
down_count = 0

# 학생들의 총 점수 계산
for score in scores:
    total += score

# 학생들의 평균 점수 계산
avg = total / len(scores)

# 학생들의 평균 점수 이상과 이하를 받은 학생 수 계산
for score in scores:
    if score >= avg:
        up_count += 1
    else:
        down_count += 1

print(f"학생들은 총 {len(scores)}명 입니다.")
print(f"학생들의 평균 점수는 {avg}점 입니다.")
print(f"평균 점수 이상의 점수를 받은 학생은 {up_count}명 입니다.")
print(f"평균 점수 이하의 점수를 받은 학생은 {down_count}명 입니다.")
````

## Part 2. 딕셔너리를 이용한 단어 빈도 분석
다음과 같은 문장을 대상으로 단어가 몇 번 등장하는지 분석합니다.

```
sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""
```

다음 과정을 직접 구현합니다.

1. `split()`으로 문장을 단어 단위로 나눕니다.
2. 빈 딕셔너리를 만듭니다.
3. 단어를 하나씩 반복합니다.
4. 처음 등장한 단어는 딕셔너리에 `1`로 저장합니다.
5. 이미 존재하는 단어는 기존 값에 `1`을 증가시킵니다.
6. 전체 단어와 등장 횟수를 출력합니다.
7. 일정 횟수 이상 등장한 단어만 별도로 출력합니다.

예:

```
파이썬 : 4회
공부는 : 2회
...
```

**필수 조건**

- `for` 사용
- `if` 사용
- `dict` 사용
- `collections.Counter` 등 자동 빈도 계산 기능은 사용하지 않음

**필수 역량**

`str`, `split()`, `dict`, `for`, `if`

````python
sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""

words = sentence.split()

word_count = {}

for word in words:
    if word in word_count:
        word_count[word] = word_count[word] + 1
    else:
        word_count[word] = 1

# 전체 단어와 각 단어의 등장 횟수 출력
print("[ 각 단어의 등장 횟수 ]")
for word in word_count:
    if word_count[word]:
        print(f"{word} : {word_count[word]}회")

# 단어가 2회 이상 등장한 단어만 출력
print("[ 2회 이상 등장한 단어 ]")
for word in word_count:
    if word_count[word] >= 2:
        print(f"{word} : {word_count[word]}회")
````

## Part 3. 함수로 반복되는 코드 정리하기
Part 2에서 작성한 단어 빈도 분석 기능을 함수로 만들어봅니다.

예:

```
defcount_words(sentence):# 단어 빈도를 계산하는 코드
    ...
```

다음 내용을 확인합니다.

- 함수에 문자열 전달
- 함수 내부에서 단어 빈도 계산
- 계산 결과 반환
- 반환된 결과 출력

가능하다면 특정 횟수 이상 등장한 단어만 출력하는 기능도 별도 함수로 분리해봅니다.

**필수 역량**

`def`, 매개변수, `return`

````python
sentence = """
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""

# 각 단어의 빈도 계산 함수
def count_words(sentence):
    
    words = sentence.split()
    word_count = {}

    for word in words:
        if word in word_count:
            word_count[word] = word_count[word] + 1
        else:
            word_count[word] = 1

    return word_count

# 단어 빈도가 2회 이상인 단어만 출력하는 함수
def two_count_words(word_count):
    print("[ 2회 이상 등장한 단어 ]")
    for word in word_count:
        if word_count[word] >= 2:
            print(f"{word} : {word_count[word]}회")
    return

# 단어 빈도가 3회 이상인 단어만 출력하는 함수
def three_count_words(word_count):
    print("[ 3회 이상 등장한 단어 ]")
    for word in word_count:
        if word_count[word] >= 3:
            print(f"{word} : {word_count[word]}회")
    return

word_count = count_words(sentence)

print(word_count)
two_count_words(word_count)
three_count_words(word_count)
````

## Part 4. 오류 확인 및 코드 검증
작성한 코드에 문제가 발생했을 때 바로 전체 코드를 다시 만들지 않고 오류 원인을 확인합니다.

다음 순서로 확인합니다.

```
오류 메시지 확인
→ 문제가 발생한 코드 위치 확인
→ 변수 값 확인
→ 자료형 확인
→ 코드 수정
→ 다시 실행
```

가능하다면 예상 가능한 오류에 대해 `try-except`를 사용해 처리해봅니다.

````python
# 함수에 잘못된 값을 전달했을 때 발생하는 오류 생성
# 숫자 123에는 .split() 메서드가 없기 때문에 AttributeError가 발생한다.
# try-except를 사용하여 오류 발생 시 오류 메시지 출력

sentence = """
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""


# 각 단어의 빈도 계산 함수
def count_words(sentence):
    words = sentence.split()
    word_count = {}

    for word in words:
        if word in word_count:
            word_count[word] = word_count[word] + 1
        else:
            word_count[word] = 1

    return word_count

# 단어 빈도가 2회 이상인 단어만 출력하는 함수
def two_count_words(word_count):
    print("[ 2회 이상 등장한 단어 ]")
    for word in word_count:
        if word_count[word] >= 2:
            print(f"{word} : {word_count[word]}회")
    return

# 단어 빈도가 3회 이상인 단어만 출력하는 함수
def three_count_words(word_count):
    print("[ 3회 이상 등장한 단어 ]")
    for word in word_count:
        if word_count[word] >= 3:
            print(f"{word} : {word_count[word]}회")
    return

# 오류 발생 테스트
try:
    word_count = count_words(123)
except AttributeError as e:
    print("오류가 발생했습니다.")
    print(e)

# 올바른 값으로 다시 실행
word_count = count_words(sentence)

print(word_count)
two_count_words(word_count)
three_count_words(word_count)
````

## 도전 문제 - 사용자가 입력한 기준 횟수 이상 등장한 단어만 출력하기

````python
sentence = """
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""


# 각 단어의 빈도 계산 함수
def count_words(sentence):
    words = sentence.split()
    word_count = {}

    for word in words:
        if word in word_count:
            word_count[word] = word_count[word] + 1
        else:
            word_count[word] = 1

    return word_count


# 사용자가 입력한 기준 횟수 이상 등장한 단어 출력 함수
def count_over_words(word_count, count):
    print(f"[ {count}회 이상 등장한 단어 ]")

    for word in word_count:
        if word_count[word] >= count:
            print(f"{word} : {word_count[word]}회")


# 단어 빈도 계산
word_count = count_words(sentence)

# 사용자에게 기준 횟수 입력받기
count = int(input("몇 회 이상 등장한 단어를 출력할까요? : "))

# 결과 출력
count_over_words(word_count, count)
````
